# Plots the results from the C++ code

```python
import csv

import matplotlib.pyplot as plt
import numpy as np


def plot_trajectory(frame: np.ndarray, boxes1: list, boxes2: list = None):
    """Plots the trajectories with different colors

    Args:
        frame: The frame to plot on
        boxes1: First trajectory (original measurements)
        boxes2: Second trajectory (Kalman predictions) if provided
    """
    # Convert boxes to numpy arrays for plotting
    boxes1_arr = np.array(boxes1)
    
    # Plot original trajectory as a continuous line
    plt.plot(boxes1_arr[:, 0], boxes1_arr[:, 1], 'r-', label='Original')

    # Plot Kalman predictions if provided
    if boxes2 is not None:
        boxes2_arr = np.array(boxes2)
        plt.plot(boxes2_arr[:, 0], boxes2_arr[:, 1], 'b-', label='Kalman Filter')

    plt.legend()
    plt.grid(True)
    plt.xlabel('X Position')
    plt.ylabel('Y Position') 
    plt.show()
    return frame

def read_csv(file_path: str):
    with open(file_path, 'r') as file:
        reader = csv.reader(file)
        data = list(reader)
    final_data = []
    for row in data:
        final_data.append([float(x) for x in row])
    return final_data

def separate_data(data: list, first: bool = False):
    boxes = []
    for row in data:
        if first:
            boxes.append([row[1], row[2]])
        else:
            boxes.append([row[3], row[4]])
    return boxes

csv_path = "/Users/derek/Desktop/drone_visualize/kalman_filtering/build/kalman_output.csv"
data = read_csv(csv_path)
original_data = separate_data(data, first=True)
kalman_data = separate_data(data, first=False)

frame = np.zeros((100, 100))
plot_trajectory(frame, original_data, kalman_data)
```
